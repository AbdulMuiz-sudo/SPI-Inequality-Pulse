```python
%sql
SELECT * FROM workspace.default.bronze_wfp_raw LIMIT 20;
```

```python
%sql
-- Bronze WFP: raw text rows with source and batch
SELECT adm0_name, mkt_name, cm_name, mp_year, mp_month, mp_price, source_file, batch_id, load_timestamp
FROM workspace.default.bronze_wfp_raw WHERE adm0_name = 'Pakistan' LIMIT 5;

```

```python
%sql
-- Bronze PBS report: raw Excel cells (page 1 shows the quintile table)
SELECT week_ending_date, sheet_name, row_idx, c01, c03, c04, c06, c07, source_file, load_timestamp
FROM workspace.default.bronze_pbs_report
WHERE week_ending_date = '24.09.2026' AND sheet_name = 'Page 1' ORDER BY row_idx LIMIT 10;

```

```python
%sql
-- Bronze PBS annex: raw Excel cells (wide city layout)
SELECT week_ending_date, sheet_name, row_idx, c00, c01, c02, c03, c04, c05, source_file
FROM workspace.default.bronze_pbs_annex
WHERE week_ending_date = '24.09.2026' ORDER BY sheet_name, row_idx LIMIT 10;

```

```python
%sql
-- Silver WFP: typed prices, one row per market, commodity and month
SELECT country, market, commodity_name, category, price_type, price, currency, price_date, is_revised_flag
FROM workspace.default.silver_wfp_price_observation ORDER BY price_date DESC LIMIT 5;

```

```python
%sql
-- Silver quintile index: 6 rows per week (Q1 to Q5 + Combined)
SELECT week_ending_date, quintile, income_band, index_value, pct_change_wow, pct_change_yoy
FROM workspace.default.silver_quintile_index
WHERE week_ending_date = (SELECT max(week_ending_date) FROM workspace.default.silver_quintile_index)
ORDER BY quintile;

```

```python
%sql
-- Silver item weights: one row per item per week
SELECT week_ending_date, item_name, price_direction, pct_change_wow, weight_lowest, impact_lowest
FROM workspace.default.silver_item_weight_impact
WHERE week_ending_date = (SELECT max(week_ending_date) FROM workspace.default.silver_item_weight_impact)
ORDER BY impact_lowest DESC LIMIT 5;

```

```python
%sql
-- Silver city prices: one row per item per city per week
SELECT week_ending_date, item_name, city, min_price, avg_price, max_price, price_dispersion_pct
FROM workspace.default.silver_city_price_observation
WHERE week_ending_date = (SELECT max(week_ending_date) FROM workspace.default.silver_city_price_observation)
ORDER BY item_sl, city LIMIT 5;

```
